**Summary of the dataset**

I calculated descriptive statistics for the 4 numeric columns in the Iris dataset — sepal_length, sepal_width, petal_length, and petal_width.

-For sepal_length and sepal_width, the mean, median, and mode values are all quite close to each other. This shows that their distribution is roughly symmetric, without any major skew in these two columns. The standard deviation is also low (0.83 and 0.44), meaning the values stay fairly close to their average.

-For petal_length and petal_width, I noticed a bigger difference between the mean, median, and mode. For example, petal_length had a mean of 3.76 but a mode of only 1.4 — at first this looked like the data was skewed. But when I grouped the data by species using groupby('species'), it became clear that there are actually 3 different species with very different petal sizes:

-Setosa has the smallest petals (length ~1.46, width ~0.25)
Versicolor is medium-sized (length ~4.26, width ~1.33)
Virginica has the largest petals (length ~5.55, width ~2.03)

**-Looking at the 25th and 75th percentiles also confirms this — the middle 50% of petal_length values (1.6 to 5.1) is a much wider range compared to sepal_width (2.8 to 3.3), which shows petal measurements are far more spread out across the different species.**

-So the gap wasn't due to skewness in a single distribution — it was because data from three different groups was mixed together.

My takeaway: If mean, median, and mode differ a lot for a column, it's not always correct to assume the data is skewed. It's important to first check whether there's a hidden category or group in the data causing that pattern. In the Iris dataset, species turned out to be exactly that hidden factor.
